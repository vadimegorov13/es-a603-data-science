```python
import pandas as pd

df = pd.read_csv('../../data/lung_cancer_dataset.csv')
```

```python
df
```

```python
cancer = df[df['lung_cancer'] == "Yes"]["age"]
no_cancer = df[df['lung_cancer'] == "No"]["age"]
```

```python
print("Mean age, cancer:", cancer.mean())
print("Mean age, no cancer:", no_cancer.mean())
```

```python
from scipy.stats import ttest_ind

t_stat, p_value = ttest_ind(cancer, no_cancer, equal_var=False)
print("T-statistic:", t_stat)
print("P-value:", p_value) # Tells us that there is a significant difference between the two groups. The p-value is less than 0.05, which means we can reject the null hypothesis that there is no difference in mean age between those with and without lung cancer.
```

```python
# what proportion of the patients have a family history of lung cancer?

family_history_proportion = df['family_history'].value_counts()
print("Proportion of patients with family history of lung cancer:", family_history_proportion)
```

```python
import numpy as np

p0 = 0.5 # Null hypothesis proportion
n = len(df) # Sample size

SE = np.sqrt(p0 * (1 - p0) / n)

print("Standard Error:", SE)
```

```python
p_hat = family_history_proportion['Yes'] / n
z_stat = (p_hat - p0) / SE
print("Z-Statistic:", z_stat)
```

```python
table = pd.crosstab(df['lung_cancer'], df['family_history'])
table
```

```python
from scipy.stats import chi2_contingency

chi2, p_value, dof, expected = chi2_contingency(table)
print("Chi-squared Statistic:", chi2)
print("P-value:", p_value) # Tells us that there is a significant association between
print("Degrees of Freedom:", dof)
print("Expected Frequencies:")
print(expected)


# What is the null hypothesis for the chi-squared test?
# A: The null hypothesis for the chi-squared test is that there is no association between lung cancer and family history of lung cancer. In other words, the two variables are independent of each other.
```

```python
if p_value < 0.05:
    print("Reject the null hypothesis: There is a significant association between lung cancer and family history of lung cancer.")
else:
    print("Fail to reject the null hypothesis: There is no significant association between lung cancer and family history of lung cancer.")
```
